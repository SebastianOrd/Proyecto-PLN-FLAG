```python
from transformers import AutoTokenizer, AutoModelForCausalLM

BASE = "deepseek-ai/deepseek-coder-1.3b-instruct"

model = AutoModelForCausalLM.from_pretrained(BASE, trust_remote_code=True)
tokenizer = AutoTokenizer.from_pretrained(BASE, use_fast=True, trust_remote_code=True)

model.save_pretrained("src/models/deepseek-coder-1p3b-lora-base")
tokenizer.save_pretrained("src/models/deepseek-coder-1p3b-lora-base")

print("Modelo base correcto descargado.")
```
